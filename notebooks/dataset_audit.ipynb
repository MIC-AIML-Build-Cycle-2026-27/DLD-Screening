## Interpretation and combination decision

No age model can be evaluated on the acquired audio. Never fill Zenodo ages with its cohort mean or borrow Ohio ages. A source–age relationship is not estimable because ages and audio belong to different cohorts. Prefer one OCSC task after access and ID joins. Split children before balancing, fitting imputers/scalers or feature selection; keep every session, clip and microphone variant of each child together. Evaluate mean-age, acoustic, linguistic and combined models on the same held-out children, with MAE/RMSE and child-level uncertainty. Do not interpret predicted-age gaps as DLD risk.
